```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.lines import Line2D

# File paths for the two plots
a_plot_file = 'figures/mike-made/seeding.bathy.iceshelves.png'
b_plot_file = 'figures/maps/Reference-Map-recentered.png'

# Load the images
a_img = mpimg.imread(a_plot_file)
b_img = mpimg.imread(b_plot_file)
```

```python
# Save the initial figure
save_path = 'figures/manuscript-versions/fig_4_1_reference-map.png'

# Create a figure for the 2-panel layout
fig, axs = plt.subplots(1, 2, figsize=(18, 10), constrained_layout=True)

# Plot the images
axs[0].imshow(a_img, aspect='equal')
axs[0].axis('off')  # Remove axis for image A
axs[0].set_title('A', loc='left', fontsize=20, fontweight='bold')

axs[1].imshow(b_img, aspect='equal')
axs[1].axis('off')  # Remove axis for image B
axs[1].set_title('B', loc='left', fontsize=20, fontweight='bold')

# Add a legend under Plot A with an empty triangle marker
legend_elements = [ Line2D([0], [0], marker='^', color='blue', markerfacecolor='none',
                            linestyle='None', markersize=10, markeredgewidth=3, label='Release Location') ]
axs[0].legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.5, 0.02), fontsize=14, frameon=False)

plt.savefig(save_path, dpi=300, bbox_inches='tight')

# Display the figure for testing
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.gridspec import GridSpec
from matplotlib.lines import Line2D

# Load and rotate image A by 90 degrees clockwise
a_img = np.rot90(mpimg.imread('figures/mike-made/seeding.bathy.iceshelves.png'), k=-1)

# Load image B
b_img = mpimg.imread('figures/maps/Reference-Map-recentered.png')

# Get the aspect ratios (width / height)
a_aspect = a_img.shape[1] / a_img.shape[0]
b_aspect = b_img.shape[1] / b_img.shape[0]

# Ratio of the widths needed to match heights
width_ratio = a_aspect / b_aspect

# Create a figure with constrained layout and unequal widths
fig = plt.figure(figsize=(18, 10), constrained_layout=True)
gs = GridSpec(1, 2, figure=fig, width_ratios=[width_ratio, 1])

# First subplot (A)
ax0 = fig.add_subplot(gs[0])
ax0.imshow(a_img)
ax0.axis('off')
ax0.set_title('A', loc='left', fontsize=20, fontweight='bold')
legend_elements = [Line2D([0], [0], marker='^', color='blue', markerfacecolor='none',
                          linestyle='None', markersize=10, markeredgewidth=3, label='Release Location')]
ax0.legend(handles=legend_elements, loc='upper center', bbox_to_anchor=(0.5, 0.02), fontsize=14, frameon=False)

# Second subplot (B)
ax1 = fig.add_subplot(gs[1])
ax1.imshow(b_img)
ax1.axis('off')
ax1.set_title('B', loc='left', fontsize=20, fontweight='bold')

# Save
save_path = 'figures/manuscript-versions/fig_4_1_reference-map.png'
plt.savefig(save_path, dpi=300, bbox_inches='tight')
plt.show()
```
