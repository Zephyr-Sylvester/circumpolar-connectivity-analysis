```python
import os

# Path to your folder
folder = 'figures/full-res-figs'

# Print the file names and their sizes
print("Filename".ljust(50) + "Size (KB)")
print("-" * 60)
for file in os.listdir(folder):
    file_path = os.path.join(folder, file)
    if os.path.isfile(file_path):  # Only process files, not subdirectories
        file_size_kb = os.path.getsize(file_path) / 1024  # Convert bytes to KB
        print(f"{file.ljust(50)} {file_size_kb:.2f}")
```

```python
from PIL import Image
import os

folder = 'figures/full-res-figs'
compressed_folder = 'figures/compressed'

# Ensure the compressed folder exists
os.makedirs(compressed_folder, exist_ok=True)

for file in os.listdir(folder):
    file_path = os.path.join(folder, file)
    if os.path.isfile(file_path) and file.endswith(('.png', '.jpg', '.jpeg')):
        with Image.open(file_path) as img:
            # Reduce resolution to 80% and save with optimized settings
            img = img.resize((int(img.width * 0.8), int(img.height * 0.8)))
            compressed_path = os.path.join(compressed_folder, file)
            img.save(compressed_path, optimize=True, quality=85)  # Adjust quality as needed
            print(f"Compressed {file} -> {compressed_path}")
```

```python
from PIL import Image
import os

folder = 'figures/compressed'
resized_folder = 'figures/resized'

os.makedirs(resized_folder, exist_ok=True)

for file in os.listdir(folder):
    file_path = os.path.join(folder, file)
    if os.path.isfile(file_path) and file.endswith(('.png', '.jpg', '.jpeg')):
        with Image.open(file_path) as img:
            # Set target width/height (adjust as needed)
            max_width = 1920  # Example width for resizing
            max_height = 1080
            
            img.thumbnail((max_width, max_height))  # Maintain aspect ratio
            resized_path = os.path.join(resized_folder, file)
            img.save(resized_path, optimize=True, quality=85)
            print(f"Resized {file} -> {resized_path}")
```

```python
for file in os.listdir(folder):
    file_path = os.path.join(folder, file)
    if file.endswith('.png'):  # Only convert PNG files
        with Image.open(file_path) as img:
            jpeg_path = os.path.join(resized_folder, file.replace('.png', '.jpg'))
            img = img.convert('RGB')  # PNG to JPEG conversion requires RGB
            img.save(jpeg_path, "JPEG", quality=85)  # Adjust quality as needed
            print(f"Converted {file} to JPEG -> {jpeg_path}")
```

```python
total_size = 0
for file in os.listdir(resized_folder):
    file_path = os.path.join(resized_folder, file)
    if os.path.isfile(file_path):
        total_size += os.path.getsize(file_path)
print(f"Total size of all figures: {total_size / 1024:.2f} KB")
```
