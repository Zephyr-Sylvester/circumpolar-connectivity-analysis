```python
import os
from PIL import Image
```

```python
# File Paths to data
repo_path = '/Users/zephyrsylvester/repos/circumpolar-connectivity-analysis/'
fig_path = repo_path + 'figures/path-plots/'

plot_folder = repo_path + "fig-lib"  # Adjust if needed
output_folder = "gifs"   # Folder to save GIFs
# os.makedirs(output_folder, exist_ok=True)  # Create if it doesn't exist
```

```python
# Months order: 11, 12, 1, 2, 3, 4, 5, 6
month_order = [11, 12, 1, 2, 3, 4, 5, 6]

def extract_month(filename):
    """Extract the month from the filename."""
    # Assuming the filename contains "month[XX]", extract the month as an integer
    start = filename.find("month[") + 6
    end = filename.find("]", start)
    return int(filename[start:end])

def create_gif_for_year(location, year, hyp):
    """Create a GIF for the specified location and year."""
    images = []

    # Collect all images matching the location and year
    for root, _, files in os.walk(plot_folder):
        for file in files:
            if location in file and str(year) in file and file.endswith(".png"):
                images.append(os.path.join(root, file))

    # Sort images by month
    images.sort(key=lambda x: month_order.index(extract_month(x)))

    # Load images into a list of PIL Image objects
    frames = [Image.open(img) for img in images]

    # Save the frames as a GIF
    gif_path = f"{output_folder}/{location}_{year}_{hyp}.gif"
    frames[0].save(
        gif_path, save_all=True, append_images=frames[1:], 
        duration=500, loop=0
    )
    print(f"GIF saved at {gif_path}")
```

```python
plot_folder = 'fig-lib/transport-by-ng/routes-by-age-and-loc-h_ice/'

# Generate GIFs for location 'BS' for 2016, 2017, and 2018
location = "BS"
for year in [2016, 2017, 2018]:
    create_gif_for_year(location, year)
```

```python
# Which hypothesis?
hyp = 'h_size'
plot_folder = f'fig-lib/transport-by-ng/routes-by-age-and-loc-{hyp}/'

# List of locations
locations = ["BS", "GERL", "GP", "MB2"]

# Generate GIFs for all locations and years
for location in locations:
    for year in [2016, 2017, 2018]:
        create_gif_for_year(location, year, hyp)
```
